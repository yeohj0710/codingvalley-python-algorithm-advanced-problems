````python
# 완성 코드
graph = {
    "본관": ["별관", "도서관"],
    "별관": ["본관", "도서관", "체육관"],
    "강당": ["기숙사"],
    "도서관": ["본관", "별관"],
    "기숙사": ["강당"],
    "체육관": ["별관"],
}
visited = set()


def dfs(node):
    visited.add(node)
    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs(neighbor)


start = "본관"
dfs(start)

for building in graph:
    if building in visited:
        print(f"{building}에는 비를 맞지 않고 도착할 수 있습니다.")
    else:
        print(f"{building}에는 비를 맞지 않고는 도착할 수 없습니다.")
````
