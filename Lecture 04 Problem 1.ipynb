### 문제

우성이는 던전을 탐험하고 있습니다. 그런데 아뿔싸, 던전이 너무 복잡한 나머지 나가는 길을 도저히 찾을 수가 없네요. 다행히 몇 번 구역이 몇 번 구역과 연결되어 있는지 적힌 단서를 발견했습니다. 다만 출구가 몇 번 구역에 있는지는 직접 찾아야 합니다. 우성이는 길을 헤매지 않기 위해 정해진 규칙에 따라 구역들을 방문하며 출구를 찾으려 합니다. 우성이를 도와 아래 조건에 맞게 던전을 탐색하는 경로를 구하는 코드를 완성해 주세요.

### 조건

- 구역들은 1번부터 차례대로 번호를 가지고 있으며, 우성이가 현재 위치한 구역의 번호는 1번입니다.
- 구역들 사이의 연결 관계는 인접 리스트로 저장하며, 인접한 구역들의 번호는 오름차순으로 저장합니다.
- 현재 구역에서 아직 방문하지 않은 구역들 중 가장 번호가 낮은 구역부터 방문합니다.
- 인접한 구역들 중 더 이상 방문하지 않은 구역이 없으면 마지막에 있었던 방으로 돌아갑니다.

### 빈칸 코드
```python
graph = {1: [2, 3], 2: [1, 4, 5], 3: [1, 6], 4: [2], 5: [2], 6: [3]}
visited = set()

def dfs(node):
    visited.□□□(node)
    print(f"{node}번 구역으로 이동합니다.")

    for neighbor in graph[node]:
        if neighbor not in □□□□□□□:
            dfs(□□□□□□□□)
            print(f"{node}번 구역으로 되돌아옵니다.")

dfs(1)
```

````python
# 완성 코드
graph = {1: [2, 3], 2: [1, 4, 5], 3: [1, 6], 4: [2], 5: [2], 6: [3]}
visited = set()


def dfs(node):
    visited.add(node)
    print(f"{node}번 구역으로 이동합니다.")

    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs(neighbor)
            print(f"{node}번 구역으로 되돌아옵니다.")


dfs(1)
````
