### 문제

동아리방의 과자를 도둑맞았습니다! 과자 도둑은 하루 사이에 거대한 유리병 크기 분량의 과자를 모두 먹어치우고 유유히 사라졌습니다. 코딩 동아리의 회장 기피와 부회장 너비는 동아리방 근처에서 목격된 친구들을 수소문해서 한 명씩 조사하려고 합니다. 조사는 최초 발견자인 지훈이부터 시작해, 목격된 사람들을 조사하는 방식으로 진행됩니다.

동아리 회장 기피는 목격된 사람을 계속해서 따라가며 깊이 우선 탐색 방식으로 조사하는 방법이 범인을 가장 빨리 찾을 것이라고 주장합니다. 하지만 동아리 부회장 너비는 목격된 사람들을 모두 먼저 조사한 뒤, 각 사람들이 목격한 사람들을 차례로 조사하는 너비 우선 탐색 방식이 가장 빠를 것이라고 주장합니다. 결국 의견 차이로 인해 기피와 너비는 각자만의 방식으로 조사를 시작하기로 했습니다. 기피의 너비를 도와 과자 도둑을 찾아주세요.

### 조건

- 각 친구들이 목격한 다른 친구들의 이름 목록은 딕셔너리 `network`에 저장되어 있습니다.
- 깊이 우선 탐색 방식과 너비 우선 탐색 방식으로 각각 범인을 찾는 과정을 출력합니다.
- 기피와 너비의 조사가 각각 끝난 후, 범인을 찾기까지 총 몇 명을 조사했는지 확인합니다.

```python
# 완성 코드
from collections import deque

network = {
    "지훈": ["석민", "민정", "지웅"], "석민": ["지훈", "유석", "지민"],
    "유석": ["석민", "예린", "민혁"], "지민": ["석민", "하진"],
    "민정": ["지훈", "도현"], "지웅": ["지훈", "수현", "서준"],
    "예린": ["유석", "서빈"], "민혁": ["유석"], "하진": ["지민"],
    "도현": ["민정"], "수현": ["지웅"], "서준": ["지웅"],
    "서빈": ["예린", "태민"], "태민": ["서빈"],
}
first_finder = "지훈"  # 최초 발견자
thief = "민혁"  # 과자 도둑의 이름
checked = set()  # 조사한 사람 목록을 저장할 집합


def dfs(person):
    checked.add(person)
    print(f"{person}이를 조사합니다.")

    if person == thief:
        print(f"과자 도둑 {thief}이의 자백을 받아냈습니다!")
        return True  # 도둑을 찾은 경우 True 반환

    print(f"{person}이는 과자 도둑이 아니었습니다.")

    for friend in network[person]:
        if friend not in checked:
            if dfs(friend):  # 현재 경로에서 도둑을 찾은 경우
                return True  # 마찬가지로 True 반환

    return False  # 현재 경로에 도둑이 없는 경우 False 반환


def bfs(start_person):
    queue = deque([start_person])
    checked.add(start_person)

    while queue:
        person = queue.popleft()
        print(f"{person}이를 조사합니다.")

        if person == thief:
            print(f"과자 도둑 {thief}이의 자백을 받아냈습니다!")
            return

        print(f"{person}이는 과자 도둑이 아니었습니다.")

        for friend in network[person]:
            if friend not in checked:
                queue.append(friend)
                checked.add(friend)


print("--- 기피의 조사 기록 ---")
dfs(first_finder)
print(f"기피는 총 {len(checked)}명을 조사했습니다.")

checked = set()  # 조사 기록 초기화
print()

print("--- 너비의 조사 기록 ---")
bfs(first_finder)
print(f"너비는 총 {len(checked)}명을 조사했습니다.")
```
