````python
# 완성 코드
items = [  # (물건 이름, 부피)
    ("책", 7), ("노트북", 19), ("스피커", 13),
    ("모니터", 23), ("고기", 5),("프린터", 17),
]
n = len(items)
max_capacity = 70  # 가방의 최대 부피

best_combination = []  # 최적의 조합을 저장할 리스트
best_volume = 0  # 가방에 담을 수 있는 최대 부피 합

def backtrack(start, selected, current_volume):
    if current_volume > max_capacity:
        return  # 가방의 최대 부피를 넘을 경우 조기에 종료
    
    global best_combination, best_volume  # 전역 변수로 사용

    if current_volume > best_volume:  # 최고 부피 합을 넘어가면 정답 갱신
        best_combination = selected
        best_volume = current_volume

    for i in range(start, n):
        item_name, item_volume = items[i]  # 물건 이름, 부피
        backtrack(i + 1, selected + [item_name], current_volume + item_volume)

backtrack(0, [], 0)

print("가방에 담을 물건 목록:", best_combination)
print("가방에 담을 수 있는 최대 부피 합:", best_volume)
````
