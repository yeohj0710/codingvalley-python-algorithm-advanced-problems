````python
# 완성 코드
class Node:
    def __init__(self, value):
        self.value = value
        self.children = []


folders = [Node(i) for i in range(7)]

folders[0].children = [folders[1], folders[2]]
folders[1].children = [folders[3], folders[4], folders[5]]
folders[2].children = [folders[6]]


def print_tree(node, level=0):
    print("  " * level + "ㄴ" + str(node.value))
    for child in node.children:
        print_tree(child, level + 1)


root = folders[0]
print_tree(root)
````
