### 문제

인태는 친구들과 영화를 관람하기 위해 영화관 좌석을 예매하려고 합니다. 예매 프로그램에서 이미 예약된 좌석은 ‘X’, 예약되지 않은 좌석은 ‘O’로 나타납니다. 인태는 최대한 많은 수의 친구들을 데리고 일렬로 예매하고 싶었지만, 중간에 예매된 자리가 많아 중간에 최대 한 자리까지는 다른 사람이 끼어 있더라도 상관 없이 가장 긴 일렬 자리를 예매하려고 합니다. 인태가 조건에 맞게 일렬로 예매할 수 있는 최대 좌석 수를 구할 수 있도록 도와주세요.

### 조건

- 좌석의 수가 n개일 때, 투 포인터 알고리즘으로 O(n) 시간에 문제를 해결합니다.
- 두 포인터는 각각 구간의 왼쪽 끝 인덱스와 오른쪽 끝 인덱스를 나타냅니다.
- 두 포인터 모두 왼쪽 끝에서 시작합니다.
- 두 포인터가 가리키는 값이 조건을 만족하는지에 따라 포인터의 위치를 유지하거나 한 칸씩 이동합니다.
    - 오른쪽 포인터를 오른쪽으로 한 칸 이동합니다.
    - 중간에 예약된 좌석이 1개보다 많을 경우 1개 이하가 될 때까지 왼쪽 포인터를 오른쪽으로 이동합니다.
- 오른쪽 포인터가 탐색 구간을 벗어날 때까지 과정을 반복합니다.

# 예약 가능한 좌석은 "O", 이미 예약된 좌석은 "X"

````python
# 완성 코드
# 예약 가능한 좌석은 "O", 이미 예약된 좌석은 "X"
seats = ["O", "O", "X", "O", "O", "X", "O", "O", "O", "X", "O"]

left = 0
max_seats = 0  # 다른 사람을 최대 한 명 포함해 일렬로 예매 가능한 좌석 수
count_reserved = 0  # 구간에 포함된 다른 사람 좌석의 수

for right in range(len(seats)):
    if seats[right] == "X":
        count_reserved += 1

    while count_reserved > 1:
        if seats[left] == "X":
            count_reserved -= 1
        left += 1

    available_seats = (right - left + 1) - count_reserved
    max_seats = max(max_seats, available_seats)

print(f"다른 사람을 최대 한 명 포함해 일렬로 예매 가능한 좌석: {max_seats}개")
````
