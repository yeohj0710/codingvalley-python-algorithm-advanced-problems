### 문제

지윤이는 뉴스를 통해 한 도시에서 전염병이 발생했다는 소식을 접하게 되었습니다. 전염병의 이름은 Blue Fever Syndrome(BFS), 푸른 반점이 생기고 열이 나는 무시무시한 질병입니다. 다행히 전염병이 발생한 도시는 지윤이가 사는 도시에서 거리가 꽤 있어, 멀리 떨어진 도시로 대피할 때까지 충분한 시간이 있습니다. 지윤이는 인접한 도시들의 관계를 알고 있으며, 인접한 도시들 사이의 거리는 모두 동일합니다. 지윤이가 전염병 BFS가 발생한 도시로부터 가장 먼 도시를 찾을 수 있도록 도와주세요.

### 조건

- 전염병이 최초로 발생한 도시는 4번 도시입니다.
- 인접한 도시들의 관계는 인접 리스트로 주어집니다.
- 인접한 도시들 사이의 거리는 모두 동일하며, 편의상 1이라고 가정합니다.

### 선택지
```
- 0
- 4
- in
- not in
- distances[node]
- distances[node] + 1
```

````python
# 완성 코드
from collections import deque

graph = {1: [2], 2: [1, 3], 3: [2, 4], 4: [3, 5, 6], 5: [4, 7], 6: [4], 7: [5]}
distances = {}


def bfs(start):
    queue = deque([start])
    distances[start] = 0

    while queue:
        node = queue.popleft()

        for neighbor in graph[node]:
            if neighbor not in distances:
                queue.append(neighbor)
                distances[neighbor] = distances[node] + 1


bfs(4)

max_city = None
max_distance = 0

for city, distance in distances.items():
    if distance > max_distance:
        max_city = city
        max_distance = distance

print(f"가장 먼 도시는 {max_city}번 도시이며, 발생지로부터의 거리는 {max_distance}입니다.")
````
