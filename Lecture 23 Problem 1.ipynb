### 문제

경자는 바이오 기업에서 일하는 연구원으로, 특정 약물을 처리했을 때 세포의 증식이 얼마나 억제되는지 확인하기 위해 세포의 개수를 세는 작업을 하고 있습니다. 어느 날 경자는 작업을 훨씬 간편하게 해 줄 획기적인 아이디어를 떠올렸는데요. 그건 바로 살아있는 세포만 푸른색으로 염색한 뒤, 파이썬을 이용해 현미경 이미지에서 푸른 덩어리의 개수를 세어 세포의 개수를 알아내는 것이었습니다.

````python
# 완성 코드
image = [
    ["R", "R", "R", "G", "G", "R", "R", "G", "B", "R"],
    ["R", "B", "B", "G", "G", "R", "R", "G", "B", "B"],
    ["R", "B", "B", "G", "R", "R", "G", "G", "B", "B"],
    ["R", "R", "R", "G", "R", "B", "B", "R", "R", "B"],
    ["G", "G", "R", "G", "R", "B", "B", "R", "R", "G"],
    ["G", "G", "R", "R", "R", "R", "R", "R", "G", "G"],
    ["R", "R", "R", "G", "R", "R", "R", "G", "G", "G"],
    ["R", "B", "B", "R", "R", "G", "G", "B", "B", "R"],
    ["R", "B", "B", "R", "G", "G", "G", "R", "B", "R"],
    ["R", "R", "R", "R", "G", "G", "R", "R", "R", "R"],
]
visited = set()
rows, cols = len(image), len(image[0])


def dfs(i, j):
    visited.add((i, j))
    for di, dj in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        ni, nj = i + di, j + dj
        if 0 <= ni < rows and 0 <= nj < cols:
            if image[ni][nj] == "B" and (ni, nj) not in visited:
                dfs(ni, nj)


count = 0

for i in range(rows):
    for j in range(cols):
        if image[i][j] == "B" and (i, j) not in visited:
            dfs(i, j)
            count += 1

print(f"세포의 개수는 총 {count}개입니다.")
````
