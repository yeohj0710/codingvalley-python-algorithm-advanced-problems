### 문제

평범한 소년 탐석이는 어느 날 기억을 모두 잃은 채 이상한 마을로 보내졌습니다. 이곳에서는 똑같은 처지의 소년들이 리더 ‘너비’를 중심으로 살아가고 있습니다. 이곳에서 벗어날 유일한 방법은 마을과 연결된 미로의 출구를 탐색하는 것입니다. 하지만 미로에서는 밤마다 괴물이 나오기 때문에, 미로가 열리는 아침에 출발해 입구가 닫히는 밤이 되기 전에 돌아와야 합니다. 미로에는 각 구역 번호가 있고, 소년들은 최대한 안전하게 탐색을 하기 위해 매일 미로 입구에서 가장 가까운 새 구역을 하나만 확인하고 돌아옵니다. ‘조건’에 소년들의 탐색 규칙이 주어질 때, 탐석이를 도와 소년들이 미로를 탐색하는 구역의 순서를 출력하는 프로그램을 완성해 주세요.

### 조건

- 소년들은 입구인 0번 구역에서 가까운 구역부터 하나씩 탐색합니다.
- 새로운 구역을 탐색하는 경우, 이 구역과 인접하면서 아직 방문하지 않은 구역의 존재는 모두 발견할 수 있습니다.
- 아직 방문하지 않은 구역의 존재를 발견한다면, 다음에 방문할 대기열의 맨 뒤에 낮은 번호부터 구역 번호를 적습니다.
- 소년들은 대기열에 적힌 구역 번호들의 순서에 따라 새로운 구역을 방문합니다.
- 미로의 모든 구역들의 연결 관계는 인접 리스트의 형태로 주어집니다.

### 선택지

````python
# 완성 코드
from collections import deque

graph = {
    0: [1, 4, 5], 1: [0, 2, 3], 2: [1, 6, 7],
    3: [1, 8], 4: [0, 9], 5: [0, 10, 11],
    6: [2, 12], 7: [2], 8: [3], 9: [4],
    10: [5], 11: [5], 12: [6, 13], 13: [12],
}
visited = set()


def bfs(start):
    queue = deque([start])
    visited.add(start)
    day = 1

    while queue:
        node = queue.popleft()
        print(f"{day}일차: {node}번 구역을 탐색합니다.")
        day += 1

        for neighbor in graph[node]:
            if neighbor not in visited:
                queue.append(neighbor)
                visited.add(neighbor)


bfs(0)
````
