### 선택지
```
- graph
- graph[curr_node]
- weight
- curr_dist + weight
- next_node, next_dist
- next_dist, next_node
```

````python
# 완성 코드
import heapq


def dijkstra(graph, start):
    distances = {node: float("inf") for node in graph}
    distances[start] = 0
    priority_queue = [(0, start)]

    while priority_queue:
        curr_dist, curr_node = heapq.heappop(priority_queue)

        for next_node, weight in graph[curr_node]:
            next_dist = curr_dist + weight

            if next_dist < distances[next_node]:
                distances[next_node] = next_dist
                heapq.heappush(priority_queue, (next_dist, next_node))

    return distances


graph = {  # (인접한 역 이름, 이동 시간(분))
    "강남역": [("신논현역", 4), ("선릉역", 3)],
    "신논현역": [("강남역", 4), ("논현역", 4), ("선정릉역", 5)],
    "선릉역": [("강남역", 3), ("선정릉역", 3)],
    "논현역": [("신논현역", 4), ("강남구청역", 3)],
    "선정릉역": [("신논현역", 5), ("선릉역", 3), ("강남구청역", 3)],
    "강남구청역": [("논현역", 3), ("선정릉역", 3)],
}

start = "강남역"
distances = dijkstra(graph, start)

for node in graph:
    print(f"{start} -> {node}까지의 최단 시간: {distances[node]}분")
````
