### 선택지

````python
# 완성 코드
players = [
    ("민수", 13), ("지훈", 8), ("서준", 7), ("민호", 5), ("우영", 2),
    ("지현", 8), ("유진", 5), ("재석", 7), ("하늘", 2), ("동원", 5),
]

n = len(players)
m = n // 2


def backtrack(start, selected):
    if len(selected) == m:
        blue_team = selected
        red_team = [player for player in players if player not in blue_team]

        blue_sum = sum(player[1] for player in blue_team)
        red_sum = sum(player[1] for player in red_team)

        if blue_sum == red_sum:
            print("블루 팀:", blue_team)
            print("레드 팀:", red_team)
            print("양팀의 능력치 합:", blue_sum)
            print()
        return

    for i in range(start, n):
        backtrack(i + 1, selected + [players[i]])


backtrack(0, [])
````
