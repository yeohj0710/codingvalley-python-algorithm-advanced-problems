````python
# 완성 코드
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, value, idx=0):
        new_node = Node(value)

        if idx == 0 or not self.head:
            new_node.next = self.head
            self.head = new_node
        else:
            current = self.head
            for _ in range(idx - 1):
                if current.next == None:
                    break
                current = current.next
            new_node.next = current.next
            current.next = new_node

    def display(self):
        current = self.head
        while current:
            print(current.value, end=" -> ")
            current = current.next
        print("None")


line = LinkedList()  # 연결 리스트 생성

line.insert("일훈", 0)  # 일훈이가 1등으로 줄을 섰습니다.
line.insert("이현", 1)  # 이현이가 2등으로 줄을 섰습니다.
line.insert("삼우", 2)  # 삼우가 3등으로 줄을 섰습니다.
line.insert("새빛", 1)  # 새빛이가 새치기를 했습니다.

line.display()  # 최종적인 줄의 형태를 출력합니다.
````
