### 문제

2050년에 세계를 정복하기 위해 지구로 침략한 ‘메모이제마이신’은 컴퓨터의 메모리를 갉아먹고 사는 무서운 세균들입니다. 이 세균들의 특징은 매 한 시간마다 그 숫자가 ‘(한 시간 전 세균 수) × 2 + (두 시간 전 세균 수)’가 된다는 것입니다. 세균이 처음 등장한 두 시간 동안에는 세균의 수가 한 마리였다고 가정할 때, 세균이 등장한지 이틀 뒤인 48시간 후에는 세균이 몇 마리가 될까요? 여러분이 코드를 완성해 그 숫자를 계산해 주세요.

### 선택지
```
- hour
- hour - 1
- hour - 2
- memo[1]
- memo[2]
- memo[hour]
```

````python
# 완성 코드
memo = {}  # 메모이제이션을 위한 전역 변수


def bacteria_count(hour):
    if hour in memo:
        return memo[hour]

    if hour <= 2:  # 처음 두 시간 동안은 세균이 한 마리
        return 1

    # 계산된 세균 수를 memo에 저장
    memo[hour] = bacteria_count(hour - 1) * 2 + bacteria_count(hour - 2)

    return memo[hour]


hours = 48
result = bacteria_count(hours)
print(f"{hours}시간 후의 세균 수: {result:,}마리")
````
