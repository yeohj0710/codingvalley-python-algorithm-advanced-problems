### 문제

우석이는 학교의 에어컨 가동 방식에 불만이 많습니다. 학교에서는 일주일 평균 기온이 30도가 넘는 주부터 에어컨을 틀어준다는 기준을 내세우지만, 일주일 평균 기온이 높지 않더라도 유독 하루 이틀만 더운 날이 있기 때문입니다. 

그러던 중 우석이는 한 가지 묘안을 생각해 냈는데요. 그건 바로 일주일을 자신에게 유리하도록 자르는 것입니다. 일요일이나 월요일을 한 주의 시작으로 하여 일주일을 나누는 것이 아닌, 수요일이나 목요일처럼 아무 날이나 시작일로 잡아 일주일을 자른다면 평균 기온이 30도가 넘어가는 주가 나올 수도 있기 때문입니다. 30일치의 기온이 주어질 때, 우석이를 도와 30일 중 가장 높은 일주일 평균 기온을 구해 주세요.

### 조건

- 누적 합 알고리즘을 이용해 모든 일주일 평균 기온을 O(n) 시간에 계산합니다.
- i + 1일째까지의 기온 누적 합은 i일째까지의 기온 누적 합에 i + 1일의 기온(i 번째 인덱스의 데이터)을 더해서 구합니다.
- i일부터 i + 6일까지의 일주일 기온 합은 i + 6일째까지의 기온 누적 합에서 i - 1일째까지의 기온 누적 합을 빼서 구합니다.
- i일부터 i + 6일까지의 일주일 평균 기온은 i일부터 i + 6일까지의 일주일 기온 합을 7로 나누어 구합니다.
- 위 과정을 가능한 모든 범위에 대해 반복하며 최대 평균 기온을 갱신합니다.

### 선택지
```
- prefix_sum[i - 1]
- prefix_sum[i]
- prefix_sum[i + 1]
- temperatures[i - 1]
- 2
- 7
```

````python
# 완성 코드
temperatures = [
    28, 29, 27, 30, 31, 32, 28,
    29, 30, 29, 28, 30, 31, 29,
    28, 29, 31, 32, 30, 28, 29,
    29, 32, 31, 30, 29, 28, 29,
    31, 28
]
n = len(temperatures)

prefix_sum = [0] * (n + 1)
for i in range(n):
    prefix_sum[i + 1] = prefix_sum[i] + temperatures[i]

max_avg = 0
for i in range(1, (n + 1) - 7):
    total = prefix_sum[i + 6] - prefix_sum[i - 1]
    avg_temp = total / 7
    max_avg = max(max_avg, avg_temp)

print(f"가장 높은 일주일 평균 기온: {max_avg:.2f}도")

if max_avg > 30:
    print("에어컨을 틀 수 있습니다.")
else:
    print("에어컨을 틀 수 없습니다.")
````
