### 문제

수열을 좋아하는 수연이는 할 일이 없을 때 수열을 손으로 직접 계산하면서 구하는 것을 좋아합니다. 하지만 피보나치수열은 숫자를 조금만 계산하려고 해도 그 크기가 빠르게 커져서 계산이 너무 어려워진다는 사실을 알게 되었는데요. 이런 문제를 해결하기 위해, 긴 수열 안에서 똑같은 구간이 발생하지 않으면서도 숫자가 너무 커지지도 않는 ‘수연수열’을 만들게 되었습니다. 수연수열의 첫 번째 숫자와 두 번째 숫자는 1이고, 세 번째 숫자부터는 이전 두 숫자의 합을 999로 나눈 값을 가집니다. 수연이는 마지막으로 수열이 잘 만들어졌는지 확인하기 위해, 수연수열의 100번째 숫자마다 하나씩을 1200번째 숫자까지 출력해 보려고 합니다. 수연이가 수열의 일부를 잘 출력할 수 있도록 빈칸에 알맞은 코드를 찾아주세요.

### 조건

- 수연수열에서 0번째 숫자는 없다고 가정합니다.

- 타뷸레이션을 이용해 수열의 1200번째 숫자까지 각각의 수를 구합니다.
- 테이블을 채워나가며 100번째마다 해당하는 순서의 숫자를 출력합니다.

### 선택지
```
- n == 1
- n <= 2
- 1, n + 1
- 3, n + 1
- table[i - 2]
- tabulation(i - 2)
```

````python
# 완성 코드
def tabulation(n):
    if n <= 2:
        return 1

    table = [0] * (n + 1)
    table[1] = table[2] = 1  # 1번째 항과 2번째 항

    for i in range(3, n + 1):
        table[i] = (table[i - 1] + table[i - 2]) % 999

        if i % 100 == 0:
            print(f"수연수열의 {i}번째 항: {table[i]}")


n = 1200
tabulation(n)
````
