### 문제

유선이는 요즘 할 일이 많아지니 아예 일을 시작할 의욕을 잃어버리고 말았습니다. 그래서 고안해 낸 방법은 자동으로 할 일 목록을 관리해 주는 프로그램을 만드는 건데요. 이 프로그램은 할 일들을 24시간 우선순위에 맞게 정렬을 유지해 주고, 할 일을 수행하려고 하면 우선순위가 가장 높은 할 일을 꺼내어 알려줄 것입니다. 초기 할 일들이 `task_queue` 리스트에 주어지고, 유선이가 할 일을 하나 수행했거나 새로운 할 일이 추가되었음을 알려주는 항목들이 `actions` 리스트에 시간 순서대로 주어질 때, 유선이가 할 일을 잘 수행할 수 있도록 프로그램을 완성해 주세요.

### 조건

- `task_queue` 리스트에는 초기에 할 일 목록들이 (우선순위, 할 일 이름)의 튜플 형태로 저장되어 있습니다.
- 우선순위 숫자가 작을수록 우선순위가 높은 항목입니다.
- `actions` 리스트에는 유선이가 할 일을 하나 수행했거나 새로운 할 일이 추가되었음을 알려주는 항목들이 시간 순서대로 담겨 있습니다.
    - “수행”은 할 일을 하나 수행했음을 뜻하고, “추가”는 할 일이 하나 추가되었음을 뜻합니다.
    - “추가”의 경우에는 해당 할 일의 우선순위와 할 일 이름이 튜플에 같이 주어집니다.
- heapq 모듈의 함수를 적절히 이용하여 `task_queue` 리스트를 최소 힙으로 만들어 우선순위 큐를 구현합니다.
- 시간 순서대로 일어난 항목들을 하나씩 처리하며, 우선순위 큐에 할 일을 추가하거나 우선순위 큐에서 제거합니다.

````python
# 완성 코드
import heapq

actions = [  # (할 일 추가/수행, 우선순위, 할 일)
    ("수행", None, None),
    ("추가", 2, "과제 제출"),
    ("수행", None, None),
    ("추가", 1, "중요 메일 확인"),
    ("추가", 4, "저번 과제 점수 확인"),
    ("수행", None, None),
    ("수행", None, None),
    ("수행", None, None),
    ("수행", None, None),
]

task_queue = [
    (3, "발표 자료 준비"),
    (5, "식재료 구입"),
    (6, "방 청소"),
]

heapq.heapify(task_queue)

for action, priority, task in actions:
    if action == "추가":
        heapq.heappush(task_queue, (priority, task))
        print(f"{task}하기를 할 일 목록에 추가함 (우선순위: {priority})")
    elif action == "수행" and task_queue:
        priority, task = heapq.heappop(task_queue)
        print(f"{task} 완료! (우선순위: {priority})")
````
