### 문제

알고리즘 수업 시간에 분할 정복에 대해 배운 정욱이는 1부터 n까지의 합을 구하는 새로운 아이디어를 떠올렸습니다. 그건 바로 구간의 길이를 계속해서 반으로 나누다가, 구간의 길이가 1이 되면 다시 구간을 2개씩 합치면서 합을 구하는 것입니다. 

예를 들어 1부터 4까지의 구간을 [1, 4]로 나타낸다면, [1, 4] = [1, 2] + [3, 4]로 나눌 수 있습니다. 각각의 구간을 다시 절반으로 나누어 [1, 2] = [1, 1] + [2, 2]로, [3, 4] = [3, 3] + [4, 4]로 나타낼 수 있습니다. [1, 1] = 1, [2, 2] = 2이므로 [1, 2] = 1 + 2 = 3이 되고, [3, 3] = 3, [4, 4] = 4이므로 [3, 4] = 3 + 4 = 7이 됩니다. 마지막으로 [1, 4] = [1, 2] + [3, 4] = 3 + 7이므로 10이 됩니다.

아래 조건에 따라 정욱이가 구현하고자 하는 분할 정복 코드를 완성해 주세요.

### 조건

- 재귀 함수를 이용해 분할 정복 알고리즘을 구현합니다.
- 재귀 함수는 구간의 첫 번째 숫자와 마지막 숫자를 나타내는 left, right를 매개변수로 갖습니다.
- left와 right가 동일할 경우 [left, right] 구간의 크기가 1이므로, 구간 합과 동일한 값인 left를 반환합니다.
- 그 외의 경우에는 구간을 절반씩 나누어 두 구간에 대해 재귀 호출을 합니다.
- 재귀 호출이 끝난 이후 계산된 두 구간의 합을 더해서 최종적인 구간 합을 반환합니다.

### 빈칸 코드
```python
def sum_recursive(left, right):
    if left □□ right:
        return left

    mid = (left + right) // 2

    left_sum = sum_recursive(left, mid)
    right_sum = sum_recursive(mid + 1, right)

    return left_sum + □□□□□□□□□

n = 5
result = sum_recursive(1, n)
print(f"1부터 {n}까지의 합: {result}")
```

### 선택지
```
- <
- >
- ==
- left_sum
- right_sum
- right
```

````python
# 완성 코드
def sum_recursive(left, right):
    if left == right:
        return left

    mid = (left + right) // 2

    left_sum = sum_recursive(left, mid)
    right_sum = sum_recursive(mid + 1, right)

    return left_sum + right_sum


n = 5
result = sum_recursive(1, n)
print(f"1부터 {n}까지의 합: {result}")
````
