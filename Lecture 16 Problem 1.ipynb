### 문제

태율이는 여름철 인기 휴양지인 코딩계곡(coding valley)의 안전관리 담당자입니다. 등산로마다 난이도는 다양하지만, 어떤 등산로는 가파른 내리막길을 포함하고 있기 때문에 법적으로 위험 표지판을 설치해야 하는데요. 어떤 구간의 시작점과 끝점이 있을 때, 구간 시작점의 높이보다 끝점의 높이가 10m 이상 낮은 구간이 있으면 그 등산로에는 위험 표지판을 설치해야 합니다.

다행히도 등산로의 각 지점에서 높이를 측정한 데이터가 있었는데요. 각 지점의 데이터는 이전 지점과 비교했을 때 높이의 변화량을 뜻합니다. 예를 들어 어떤 지점의 데이터가 -3이라면, 이전 지점에 비해 3m가 더 낮은 내리막길임을 뜻합니다. 등산로의 데이터가 주어질 때, 태율이를 도와 주어진 등산로가 위험 표지판을 설치해야 하는 등산로인지 판별해 주세요.

### 조건

- 등산로는 일방통행이라고 가정합니다. 다시 말해, 등산로를 끝에서부터 거꾸로 산책할 수는 없습니다.
- 등산로 높이 데이터의 첫 번째 값은 0으로 주어집니다.
- 데이터의 수가 n개일 때, O(n) 시간에 해결할 수 있는 방법을 이용합니다.
    - 동적 계획법을 사용해 i 번째 칸에 ‘오른쪽 끝이 i 번째 숫자인 구간의 합들 중에서 가장 작은 값’을 구합니다.
    - 이전 지점의 최소 구간 합이 0보다 작은 경우에만 현재 지점의 최소 구간 합에 더하는 방식으로 테이블을 채울 수 있습니다.
    - 테이블을 모두 채운 후, 테이블에서 가장 작은 값이 가장 큰 내리막 구간의 높이 변화가 됩니다.
    - 가장 큰 내리막 구간의 높이 변화가 -10m 이하이면 위험 표지판을 설치합니다.

### 선택지
```
- 1, n
- 1, n + 1
- dp[i - 1]
- height_changes[i - 1]
- dp[n - 1]
- min(dp)
```

````python
# 완성 코드
height_changes = [0, 3, -1, 5, -2, 4, -4, 1, -4, 2, 1, -3, 1, -4, 1]
n = len(height_changes)

dp = [0] * n
dp[0] = height_changes[0]

for i in range(1, n):
    dp[i] = height_changes[i] + min(dp[i - 1], 0)

min_height = min(dp)
print(f"가장 큰 내리막 구간의 높이 변화: {min_height}m")

if min_height <= -10:
    print("이 등산로는 위험 표지판을 설치해야 합니다.")
else:
    print("이 등산로는 위험 표지판을 설치할 필요가 없습니다.")
````
