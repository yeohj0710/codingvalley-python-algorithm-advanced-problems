### 문제

동직이는 최근 코딩 강의를 아주 열심히 듣고 있습니다. 그건 바로 강의를 하나 들을 때마다 랜덤으로 받게 되는 정사각형 카드로 ‘정사각형 블랙잭’이라는 게임에 참여할 수 있기 때문인데요. 정사각형 블랙잭은 각 카드에 그려진 정사각형들의 넓이의 합이 123을 초과하지 않으면서, 123에 최대한 가깝게 만드는 게임입니다. 랜덤으로 주어지는 각 카드의 정사각형 넓이는 1, 4, 9, 16, …과 같은 제곱수 중 하나이고, 합을 구성할 때 중복되는 카드를 여러 장 사용할 수 있습니다. 넓이의 합이 123 이하인 참가자들 중에서는 합이 123에 가장 가까운 참가자의 순위가 더 높고, 같은 점수를 구성한 참가자들 중에서는 사용된 블럭의 개수가 적은 참가자가 더 높은 순위를 차지합니다.

동직이는 4개의 정사각형 카드를 가지고 (9 × 9) + (5 × 5) + (4 × 4) + (1 × 1) = 81 + 25 + 16 + 1 = 123으로 123점을 만드는 데 성공했습니다. 하지만 동직이의 목표는 이벤트 참가자들 중 오직 1등에게만 증정되는 아이패드였는데요. 그래서 남은 이벤트 기간동안 자신의 성적을 뛰어넘는 사람이 나올 수 있는지 확인하고 싶어합니다. 동직이를 도와 4개보다 더 적은 개수의 정사각형 카드로도 123을 만드는 사람이 나올 수 있는지 확인해 주세요.

### 조건

- 목표 합 123을 만들기 위한 제곱수의 최소 개수를 동적 계획법으로 구합니다.
- DP 테이블에 1부터 123까지의 각 합을 만들기 위한 제곱수의 최소 개수를 저장합니다.
- 각 합에 대해 가능한 모든 제곱수를 이용해 최소 요소 개수를 갱신합니다.
    - 현재 제곱수의 크기가 square일 때, 합 i - square를 만드는 최소 제곱수의 개수는 dp[i - square]입니다.
    - 목표 합 i를 만드는 한 가지 방법은 크기가 i - square인 합에 크기가 square인 제곱수를 합하는 것입니다.
    - 이때 사용되는 제곱수의 개수는 dp[i - square] + 1이므로, 이 값이 dp[i]보다 작으면 dp[i]를 갱신합니다.

### 빈칸 코드

### 선택지
```
- 0
- 1
- target
- target + 1
- dp[i - 1]
- dp[i - square]
```

````python
# 완성 코드
target = 123

dp = [float("inf")] * (target + 1)
dp[0] = 0

for i in range(1, target + 1):
    j = 1
    while j * j <= i:
        square = j * j
        dp[i] = min(dp[i], dp[i - square] + 1)
        j += 1

print(f"{target}을 제곱수의 합으로 표현할 때 최소 항의 개수: {dp[target]}개")

if dp[target] >= 4:
    print("동직이에게 최소 공동 1등은 보장됩니다!")
else:
    print("동직이가 2등으로 밀려날 수 있습니다.")
````
