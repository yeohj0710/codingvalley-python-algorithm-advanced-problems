### 문제

서울에 사는 정선이는 명절마다 대전에 계신 외할머니와 부산에 계신 친할머니를 뵈러 내려갑니다. 마침 최근에 그래프에 대해 공부한 정선이는 서울, 대전, 부산 사이의 KTX 철도의 연결 관계를 그래프로 나타내어 보려고 하는데요. 3개의 KTX 역을 각각의 노드로 가지고, 역 사이의 이동 시간을 가중치로 가지는 그래프를 인접 리스트로 저장하고 이를 출력하려고 합니다. 정선이를 도와 조건에 맞는 코드를 완성해 주세요.

### 조건

- **서울**에서 **대전**까지는 KTX로 **1시간**이 소요되며, 돌아올 때도 동일한 시간이 소요됩니다.
- **서울**에서 **부산**까지는 KTX로 **2시간 30분**이 소요되며, 돌아올 때도 동일한 시간이 소요됩니다.
- **대전**에서 **부산**까지는 KTX로 **1시간 30분**이 소요되며, 돌아올 때도 동일한 시간이 소요됩니다.
- 지도에 존재하는 모든 역과 역 사이의 연결 관계를 차례대로 출력합니다.

### 빈칸 코드
```python
graph = {
    "서울": [("대전", "1시간"), ("부산", "2시간 30분")],
    "대전": [("서울", "1시간"), ("부산", "1시간 30분")],
    "부산": [("서울", "2시간 30분"), ("대전", "□□□□□□□")],
}

for node in □□□□□:
    for neighbor, travel_time in graph[□□□□]:
        print(f"{node}에서 {neighbor}까지 {travel_time}이 소요됩니다.")
```

### 선택지
```
- 1시간
- 1시간 30분
- node
- neighbor
- travel_time
- graph
```

````python
# 완성 코드
graph = {
    "서울": [("대전", "1시간"), ("부산", "2시간 30분")],
    "대전": [("서울", "1시간"), ("부산", "1시간 30분")],
    "부산": [("서울", "2시간 30분"), ("대전", "1시간 30분")],
}

for node in graph:
    for neighbor, travel_time in graph[node]:
        print(f"{node}에서 {neighbor}까지 {travel_time}이 소요됩니다.")
````
