### 문제

두호는 유명 MMORPG 게임의 한 길드를 대표하는 유저입니다. 최근 두호의 길드원들이 참여할 수 있는 특별한 보스 몬스터 이벤트가 열렸습니다. 이 보스 몬스터에는 평균 레벨이 100 이상인 두 길드원이 2인 파티를 구성하여 하루에 한 번 도전할 수 있으며, 보스를 물리치면 두 길드원이 귀중한 보상을 받을 수 있습니다. 문제는 두호의 길드 중 상당수의 유저들은 레벨이 100 미만이었다는 것입니다. 하지만 마음씨 착한 두호는 최대한 많은 길드원들이 보상을 받을 수 있도록 평균 레벨이 100 이상인 파티를 최대한 많이 구성해 주려고 합니다. 두호를 도와 빈칸에 알맞은 코드를 찾아주세요.

### 조건

- 길드원들의 레벨은 `levels` 리스트에 레벨 순으로 저장되어 있습니다.
- 각 길드원은 하루에 한 번만 보스에 도전할 수 있습니다.
- 주어진 길드원 레벨 리스트에서 가능한 한 많은 2인 파티를 구성해야 합니다.
- 파티의 평균 레벨은 100 이상이어야 합니다.

### 빈칸 코드
```python
levels = [85, 87, 94, 95, 96, 103, 104, 106, 110]

left, right = 0, len(levels) - 1
parties = 0

while left < right:
    if (levels[left] + levels[right]) / 2 >= 100:
        parties += 1
        left □□□□
        right □□□□
    else:
        □□□□ += 1

print(f"구성 가능한 최대 파티 수는 {parties}개입니다.")
print(f"총 {parties * 2}명의 길드원들이 보상을 받을 수 있습니다.")
```

### 선택지
```
- left
- right
- += 1
- -= 1
- = 0
- = len(levels) - 1
```

````python
# 완성 코드
levels = [85, 87, 94, 95, 96, 103, 104, 106, 110]

left, right = 0, len(levels) - 1
parties = 0

while left < right:
    if (levels[left] + levels[right]) / 2 >= 100:
        parties += 1
        left += 1
        right -= 1
    else:
        left += 1

print(f"구성 가능한 최대 파티 수는 {parties}개입니다.")
print(f"총 {parties * 2}명의 길드원들이 보상을 받을 수 있습니다.")
````
