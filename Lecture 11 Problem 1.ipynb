````python
# 완성 코드
class Node:
    def __init__(self, value):
        self.value = value
        self.children = []


root = Node("김 부장")
root.children = [Node("이 대리"), Node("박 대리")]  # 김 부장의 직속 부하
root.children[0].children = [Node("최 사원"), Node("정 사원")]  # 이 대리의 직속 부하
root.children[1].children = [Node("강 사원"), Node("조 사원")]  # 박 대리의 직속 부하


def pre_order(node):
    print(node.value, end=" -> ")
    for child in node.children:
        pre_order(child)


def in_order(node):
    if len(node.children) > 0:
        in_order(node.children[0])

    print(node.value, end=" -> ")

    if len(node.children) == 2:
        in_order(node.children[1])


def post_order(node):
    for child in node.children:
        post_order(child)
    print(node.value, end=" -> ")


print("[김전위 씨가 서명을 받는 순서]")
pre_order(root)
print("끝!\n")

print("[이중위 씨가 서명을 받는 순서]")
in_order(root)
print("끝!\n")

print("[박후위 씨가 서명을 받는 순서]")
post_order(root)
print("끝!\n")
````
